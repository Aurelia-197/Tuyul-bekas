```
!pip install diffusers transformers accelerate torch -q
from diffusers import TextToVideoSDPipeline
from diffusers.utils import export_to_video
import torch

print('Loading AI Tuyul Bekas...')
pipe = TextToVideoSDPipeline.from_pretrained('damo-vilab/modelscope-damo-text-to-video-synthesis', torch_dtype=torch.float16)
pipe.to('cuda')

# PROMPT ANTI NGACO BOS
prompt = "cute green tuyul doll in bamboo basket waving hand, funny market background, ibu jawa jualan"
video_frames = pipe(prompt, num_frames=16).frames
export_to_video(video_frames, 'hasil_tuyul.mp4')
print('SELESAI! Video jadi: hasil_tuyul.mp4')
```
